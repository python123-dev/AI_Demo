# Research Paper Retrieval with ANNOY
Embed papers with `all-MiniLM-L6-v2`, index with ANNOY (angular, 10 trees) and find the paper most similar to a query.

```
from sentence_transformers import SentenceTransformer
from annoy import AnnoyIndex
```

```
def read_file(file_path):
    """Read a text file line by line, strip whitespace and return the lines as a list."""
    with open(file_path, "r", encoding="utf-8") as f:
        return [line.strip() for line in f if line.strip()]
```

```
def search_query(corpus, query):
    model = SentenceTransformer("all-MiniLM-L6-v2")
    embeddings = model.encode(corpus)

    dim = embeddings.shape[1]
    index = AnnoyIndex(dim, "angular")
    for i, vec in enumerate(embeddings):
        index.add_item(i, vec)
    index.build(10)

    query_embedding = model.encode(query)
    ids, distances = index.get_nns_by_vector(query_embedding, len(corpus), include_distances=True)

    most_similar_sentence = corpus[ids[0]]
    similarity_score = distances[0]
    return embeddings, index, most_similar_sentence, similarity_score
```

```
if __name__ == "__main__":
    corpus = read_file("papers.txt")
    query = "Explain about the Internet Of Things."
    embeddings, index, sentence, score = search_query(corpus, query)
    print("Query:", query)
    print("Most similar sentence:", sentence)
    print("Similarity score (angular distance):", score)
```
